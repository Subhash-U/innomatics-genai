```python
import os
from getpass import getpass

# Set your Groq API Key
if "GROQ_API_KEY" not in os.environ:
    os.environ["GROQ_API_KEY"] = getpass("Enter your Groq API Key: ")

from langchain_groq import ChatGroq
# Initialize the super-fast Llama 3.1 model
llm = ChatGroq(model="llama-3.1-8b-instant", temperature=0)
```

```python
!pip install langchain-community
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings

# 1. Initialize local open-source embeddings
embeddings = HuggingFaceEmbeddings(model_name="BAAI/bge-small-en-v1.5")

# 2. Create some dummy document chunks
documents = [
    "Refund Policy: Refunds are only processed within 30 days of purchase with a valid receipt.",
    "Password Reset: To reset your password, click 'Forgot Password' on the login screen.",
    "Shipping: Standard shipping takes 3-5 business days. Expedited shipping takes 1-2 days."
]

# 3. Load into ChromaDB
vector_store = Chroma.from_texts(documents, embeddings)
retriever = vector_store.as_retriever(search_kwargs={"k": 1})
print("Vector Store Ready!")
```

```python
from typing_extensions import TypedDict
from typing import List

class GraphState(TypedDict):
    query: str
    intent: str
    context: List[str]
    generation: str
    confidence: float
    human_override: str
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import JsonOutputParser

def router_node(state: GraphState):
    print("--- ROUTER NODE ---")
    query = state["query"]

    # Simple keyword-based intent detection (can be upgraded to LLM routing)
    escalate_words = ["agent", "human", "manager", "refund", "angry", "sue"]
    if any(word in query.lower() for word in escalate_words):
        return {"intent": "escalate"}
    return {"intent": "support"}

def retrieve_node(state: GraphState):
    print("--- RETRIEVE NODE ---")
    docs = retriever.invoke(state["query"])
    context = [doc.page_content for doc in docs]
    return {"context": context}

def generate_node(state: GraphState):
    print("--- GENERATE NODE ---")
    # Prompt asks LLM to answer AND score its own confidence (0.0 to 1.0)
    prompt = PromptTemplate(
        template="""You are a support bot. Use this context to answer the user.
        If the answer is NOT in the context, set confidence to 0.0.
        Context: {context}
        User: {query}
        Output JSON format: {{"answer": "your answer", "confidence": 0.95}}""",
        input_variables=["context", "query"]
    )

    # We enforce JSON output so we can safely extract the confidence score
    chain = prompt | llm.bind(response_format={"type": "json_object"}) | JsonOutputParser()
    result = chain.invoke({"context": state["context"], "query": state["query"]})

    return {"generation": result["answer"], "confidence": result["confidence"]}

def hitl_node(state: GraphState):
    print("---  HITL NODE (GRAPH PAUSED) ---")
    # This node acts as a placeholder. When LangGraph hits this, it pauses.
    # The human will inject the "human_override" state later.
    return state
```

```python
from langgraph.graph import StateGraph, START, END
from langgraph.checkpoint.memory import MemorySaver

# Initialize Graph
workflow = StateGraph(GraphState)

# Add Nodes
workflow.add_node("router", router_node)
workflow.add_node("retrieve", retrieve_node)
workflow.add_node("generate", generate_node)
workflow.add_node("hitl", hitl_node)

# Define Conditional Routing Logic
def route_after_intent(state: GraphState):
    if state["intent"] == "escalate": return "hitl"
    return "retrieve"

def route_after_generation(state: GraphState):
    # If confidence is below 80%, escalate to human
    if state.get("confidence", 0.0) < 0.8: return "hitl"
    return END

# Build Edges
workflow.add_edge(START, "router")
workflow.add_conditional_edges("router", route_after_intent)
workflow.add_edge("retrieve", "generate")
workflow.add_conditional_edges("generate", route_after_generation)
workflow.add_edge("hitl", END)

# Compile Graph with a Checkpointer (MemorySaver is REQUIRED for HITL pauses)
memory = MemorySaver()
app = workflow.compile(checkpointer=memory)
print("Graph Compiled Successfully!")
```

```python
# We use a thread_id to track the memory of this specific conversation
config = {"configurable": {"thread_id": "1"}}
inputs = {"query": "How do I reset my password?"}

# Run the graph
for output in app.stream(inputs, config=config):
    pass

print("\nFINAL OUTPUT:", app.get_state(config).values.get("generation"))
```

```python
# CELL 8
config = {"configurable": {"thread_id": "2"}}
inputs = {"query": "I am angry, I want a refund now!"}

# Run the graph (It will pause at the HITL node)
print("Running graph...")
for output in app.stream(inputs, config=config):
    pass

# Check where the graph is currently stopped
state = app.get_state(config)
print("\nGraph stopped at next node:", state.next) # Should print ('hitl',)

# --- THE HUMAN INTERVENES ---
print("\n[Human Agent Interface]")
print("User asked:", state.values["query"])
human_response = input("Agent, type your manual response here: ")

# Update the state with the human's response and resume the graph
app.update_state(config, {"human_override": human_response})

print("\nResuming graph...")
for output in app.stream(None, config=config):
    pass

final_state = app.get_state(config).values
print("\nFINAL DELIVERED RESPONSE:", final_state.get("human_override"))
```
